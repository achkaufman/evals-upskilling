# Use skills in Foundry

Based on [this doc](https://learn.microsoft.com/en-us/azure/foundry/agents/how-to/tools/skills?pivots=python#create-a-skill-version).

Sample skill located in this repo at /.github/skills/weather-forecast-brief.

```python
import os
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv(override=True)
```

```python
import os
from pathlib import Path
from azure.identity import DefaultAzureCredential
from azure.ai.projects import AIProjectClient
from azure.ai.projects.models import CreateSkillVersionFromFilesBody

# Create Foundry project client
endpoint = os.getenv("AZURE_AI_PROJECT_ENDPOINT")

with (
    DefaultAzureCredential() as credential,
    AIProjectClient(
        endpoint=endpoint, credential=credential, allow_preview=True
    ) as project,
):
    # Create skill version from ZIP / SKILL.md file
    skill_path = Path(".github/skills/weather-forecast-brief/SKILL.md")
    content=CreateSkillVersionFromFilesBody(
        files=[
            ("SKILL.md", skill_path.read_bytes()),
        ]
    )
    imported = project.beta.skills.create_from_files(
      "weather-forecast-brief",
      content=content,
    )
    print(
        f"Created skill: {imported.name} version: {imported.version}"
    )
```
